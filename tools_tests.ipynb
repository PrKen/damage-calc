#### Pokemon Showdown

```python
import subprocess

# Exemple de commande : simulation d'un combat
cmd = [
    "pokemon-showdown", 
    "simulate-battle"
]

# Exemple de script à transmettre au simulateur
# (pour des formats plus longs, on passerait par un input en stdin)
simulate_script = """\
>start {"formatid":"gen7randombattle"}
>player p1 {"name":"Alice"}
>player p2 {"name":"Bob"}
"""

# On lance le processus en lui passant 'simulate_script' sur l'entrée standard
process = subprocess.Popen(
    cmd,
    stdin=subprocess.PIPE,
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True  # Pour traiter le flux en texte (pas en bytes)
)

stdout_data, stderr_data = process.communicate(simulate_script)

if process.returncode == 0:
    print("Sortie du simulateur :")
    print(stdout_data)
else:
    print("Erreur dans la simulation :")
    print(stderr_data)
```

#### Pkm Dmg

```python
import subprocess

# Exemple de commande : on calcule les dégâts d'un Focus Blast d'un Gengar
# Choice Specs +1 contre une Blissey (Gen 4).
command = [
    "npx", "dmg", 
    "+1", "252", "SpA", "Gengar", "@", "Choice", "Specs", "[Focus Blast]", 
    "vs", 
    "0", "HP", "/", "172+", "SpD", "Blissey", 
    "--gen=4"
]

result = subprocess.run(command, capture_output=True, text=True)

print("=== Sortie standard (stdout) ===")
print(result.stdout)
print("=== Sortie d'erreur (stderr) ===")
print(result.stderr)
```

#### Smogon calc

```python
script_code = r"""
const {calculate, Generations, Pokemon, Move} = require('@smogon/calc');

// On choisit la 5e génération en exemple
const gen = Generations.get(5);

// On calcule le résultat d'un Focus Blast de Gengar (Choice Specs, +1 SpA)
// sur Chansey Eviolite (Calm, EVs 252/252).
const result = calculate(
  gen,
  new Pokemon(gen, 'Gengar', {
    item: 'Choice Specs',
    nature: 'Timid',
    evs: {spa: 252},
    boosts: {spa: 1}, // +1 en SpA
  }),
  new Pokemon(gen, 'Chansey', {
    item: 'Eviolite',
    nature: 'Calm',
    evs: {hp: 252, spd: 252},
  }),
  new Move(gen, 'Focus Blast')
);

console.log("=== Résultat du calcul @smogon/calc ===");
console.log("Rangs de dégâts possibles :", result.range());
console.log("Description texte         :", result.desc());
""";

# On écrit le script dans le répertoire courant
with open("test_smogon_calc.js", "w", encoding="utf-8") as f:
    f.write(script_code)

print("Fichier test_smogon_calc.js généré avec succès.")
```
